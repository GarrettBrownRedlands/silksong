**Q2** The article tells the story of a machine learning AI attempting to predict whether or not criminals would commit more crimes after they were released from jail. According to tests run by the journal, the AI, Northpointe, would incorrectly label black people to be of high-risk of reoffending nearly twice as often as it did to white people. While Northpointe is largely innacurate to begin with, the bias exemplifies this to an unreasonable level against people of color. And as much as we want to discard and ignore the prediction of a robot, it's still got some subconscious effect. Even if it's not very trustworthy, it plants the seed; a seed which might lead to someone's parole getting denied later down the line. The impacts, while subtle, can be devastating, and machine bias is at the heart of it.
